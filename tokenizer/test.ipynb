## Testing Regex Tokenizer


```python
# read all text from the file
with open("hf_tokenizer_corpus.txt", "r") as file:
    lines = file.readlines()

len(lines)
```

```python
import random

sampled_lines = random.sample(lines, 400*3)
corpus = "\n".join(sampled_lines)

corpus_bytes = corpus.encode("utf-8")
len(corpus_bytes)
```

```python
tests = [
    "आज तो बहुत थक गया हूँ, ಸ್ವಲ್ಪ विश्रಾಂತಿ ಬೇಕು।",
    "मौसम कितना अच्छा है! ನೀವೂ ಹೊರಗೆ ಬನ್ನಿ, let's enjoy together.",
    "स्वल्पा adjust करो, बैंगलोर का ट्रैफिक ऐसा ही है।",
    "ನೀವು ಚಹಾ ಕುಡಿತೀರಾ? मुझे एक cup चाहिए।",
    "आज का काम पूरा करो, ನಾಳೆ ಎಲ್ಲಿಂದ ಆರಂಭಿಸೋದು ನೋಡಿ।",
    "ಪಾರ್ಟಿ ಹೇಗೆ ಇತ್ತು? मुझे तो बहुत मजा आया!",
    "ನಮ್ಮ ಚೂರು ಸಹನಶೀಲತೆಯನ್ನು ತೋರಿಸಿ, ये थोड़ी देर का मसला है।",
    "ಸಮಯ ನಿಲ್ಲುತ್ತಿಲ್ಲ, जिंदगी में स्वल्पा मज़ा भी जरूरी है।",
    "My name is Yaseen, and I'm a software engineer."
]
```

```python
import regex as re

def test_regex(regex, text):
    print(text)
    print(re.findall(regex, text))
    print()
```

```python
regex = re.compile(
    r"""'(?i:[sdmt]|ll|ve|re)|[^\r\n\p{L}\p{N}]?+\p{L}+|\p{N}{1,3}| ?[^\s\p{L}\p{N}]++[\r\n]*|\s*[\r\n]|\s+(?!\S)|\s+"""
)

for test in tests:
    test_regex(regex, test)
```

```python
regex = re.compile(
    r"""(?i) 's|'t|'re|'ve|'m|'ll|'d| ?\b[\p{L}\u0900-\u0963|\u0966-\u097F]+\b| ?\b[\p{L}\u0C80-\u0C9E|\u0CA0-\u0CFF]+\b| ?[\p{N}]+| ?[.,!?;:'\"-]| ?[\u0964-\u0965]| ?[\u0C9E-\u0C9F]| ?[^\s\p{L}\p{N}\u0900-\u097F\u0C80-\u0CFF]+| \s+(?!\S)| \s+"""
)

for test in tests:
    test_regex(regex, test)
```

```python
import sys

sys.path.append("..")
from tokenizer.regex_tokenizer import RegexTokenizer

tokenizer = RegexTokenizer(
    regex=regex
)
```

```python
tokenizer.train(corpus, 3000, verbose=True)
```

```python
test = "hello world!!!? (안녕하세요!) lol123 😉"

def test_encoding(tokenizer, text):
    print(f"Text: {text}")
    test_ids = tokenizer.encode(text, verbose=True)
    print("")
    print(f"Unmerged length: {len(text.encode('utf-8'))}")
    print(f"Merged length: {len(test_ids)}")
    print("-"*50)

test_encoding(tokenizer, test)
```

```python
for test in tests:
    test_encoding(tokenizer, test)
```

## Testing GPT-4 Tokenizer

```python
import sys
sys.path.append("..")

# match this
import tiktoken
enc1 = tiktoken.get_encoding("cl100k_base") # this is the GPT-4 tokenizer
ids1 = enc1.encode("hello world!!!? (안녕하세요!) lol123 😉")
text1 = enc1.decode(ids1) # get the same text back


from tokenizer.gpt4 import GPT4Tokenizer
enc2 = GPT4Tokenizer()
ids2 = enc2.encode("hello world!!!? (안녕하세요!) lol123 😉", verbose=True)
text2 = enc2.decode(ids2) # get the same text back

assert ids1 == ids2
assert text1 == text2
```

```python
import sys
sys.path.append("..")

# match this
import tiktoken
enc1 = tiktoken.get_encoding("cl100k_base") # this is the GPT-4 tokenizer
ids1 = enc1.encode("<|endoftext|>hello world", allowed_special="all")
text1 = enc1.decode(ids1)

from tokenizer.gpt4 import GPT4Tokenizer
enc2 = GPT4Tokenizer()
ids2 = enc2.encode("<|endoftext|>hello world", verbose=False, allowed_special="all")
text2 = enc2.decode(ids2)

assert ids1 == ids2
assert text1 == text2
```
