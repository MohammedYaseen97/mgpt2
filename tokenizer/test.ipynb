```python
# read all text from the file
with open("hf_tokenizer_corpus.txt", "r") as file:
    lines = file.readlines()

len(lines)
```

```python
import random

sampled_lines = random.sample(lines, 400*3)
corpus = "\n".join(sampled_lines)

corpus_bytes = corpus.encode("utf-8")
len(corpus_bytes)
```

```python
import sys

sys.path.append("..")
from tokenizer.basic import BasicTokenizer

tokenizer = BasicTokenizer()
tokenizer.train(corpus, 3000, verbose=True)
```

```python
test = "hello world!!!? (안녕하세요!) lol123 😉"

def test_tokenizer(tokenizer, text):
    print(f"Text: {text}")
    test_ids = tokenizer.encode(text, verbose=True)
    print("")
    print(f"Unmerged length: {len(text.encode('utf-8'))}")
    print(f"Merged length: {len(test_ids)}")
    print("-"*100)

test_tokenizer(tokenizer, test)
```
