### Training Data for Multilingual BPE Tokenizer

Language distribution in the training corpus used to build the tokenizer:

| Language/Script | Percentage |
|----------------|------------|
| English        | 60%        |
| Hindi (Deva)   | 10%        |
| Hindi (Latn)   | 10%        |
| Kannada (Knda) | 10%        |
| Kannada (Latn) | 10%        |


```python
from datasets import load_dataset

# Load a dataset with streaming and shuffle
def load_random_subset(dataset_name, data_dir=None, name=None, split="train", num_rows=1000):
    dataset = load_dataset(dataset_name, data_dir=data_dir, name=name, split=split, streaming=True)
    shuffled_dataset = dataset.shuffle(buffer_size=10000, seed=42)  # Shuffle with a buffer
    return [next(iter(shuffled_dataset)) for _ in range(num_rows)]

# Load random subsets of each dataset using streaming
dataset_en = load_random_subset("HuggingFaceFW/fineweb-edu", name="sample-10BT", split="train", num_rows=12000)
dataset_hin_deva = load_random_subset("ai4bharat/sangraha", data_dir="synthetic/hin_Deva", split="train", num_rows=2000)
dataset_hin_latn = load_random_subset("ai4bharat/sangraha", data_dir="synthetic/hin_Latn", split="train", num_rows=2000)
dataset_kan_knda = load_random_subset("ai4bharat/sangraha", data_dir="synthetic/kan_Knda", split="train", num_rows=2000)
dataset_kan_latn = load_random_subset("ai4bharat/sangraha", data_dir="synthetic/kan_Latn", split="train", num_rows=2000)

# Print the first row of each subset to verify
print(dataset_en[0])
print(dataset_hin_deva[0])
print(dataset_hin_latn[0])
print(dataset_kan_knda[0])
print(dataset_kan_latn[0])
```
